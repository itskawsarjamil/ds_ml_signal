```python
df=pd.read_csv('./datasets/salary_data.csv')
df.dropna(inplace=True)
df.isna().sum()
x=df.drop(columns=['Salary'])
y=df['Salary']
col_name=["Years of Experience","Age","Education Level"]
col=df[["Years of Experience","Age","Education Level"]].copy()

col
```

```python
sns.barplot(x=col["Education Level"],y=y)
plt.show()
```

```python
encoder=OrdinalEncoder(categories=[["Bachelor's", "Master's", 'PhD']])
col['Education Level']=encoder.fit_transform(col[['Education Level']])
sns.barplot(x=col["Education Level"],y=y)
plt.show()
```

```python
sns.boxplot(x=col["Education Level"])

plt.show()

plt.scatter(x=col["Years of Experience"],y=y,label="exp vs sal")
plt.scatter(x=col["Age"],y=y, label="age vs sal")
plt.scatter(x=col["Education Level"],y=y, label="education_lvl vs sal")
plt.legend()
plt.show()
```
