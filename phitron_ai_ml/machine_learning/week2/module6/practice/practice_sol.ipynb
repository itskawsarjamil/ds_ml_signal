```python
import pandas as pd
import seaborn as sns
```

```python
df=pd.DataFrame(data={"Age":
                      [18, 19, 20, 20, 21, 21, 22, 22, 23, 24, 25, 26, 27, 28, 30, 32, 35, 40, 45, 70]})
age_mean=df['Age'].mean()
age_std=round(df['Age'].std(),2)
age_std
df['z_score']=(df['Age']-age_mean)/age_std
df[abs(df['z_score'])<=3].count()
```

```python
df=pd.DataFrame(data={'salaries':[25, 28, 30, 32, 32, 33, 35, 35, 36, 38, 40, 42, 45, 48, 50, 55, 60, 65, 70, 80, 120, 150, 200, 250, 500]})
Q1=df['salaries'].quantile(0.25)
Q3=df['salaries'].quantile(0.75)
IQR=Q3-Q1
salary_min=max(0,Q1-(1.5*IQR))
salary_max=(Q3+(1.5*IQR))
df[(df['salaries']<salary_min)|(df['salaries']>salary_max)]
df['salaries'].clip(salary_min,salary_max)
```

```python
dfa=pd.DataFrame(data={'scores':[55, 58, 60, 62, 65, 65, 68, 70, 72, 75, 78, 80, 85, 90, 95, 98, 100, 100, 100, 15, 20]})
dfb=pd.DataFrame(data={'prices':[30, 32, 35, 38, 40, 42, 45, 48, 50, 55, 60, 65, 70, 80, 90, 100, 120, 150, 200, 300, 400, 500, 800, 1000, 1200]})
sns.histplot(data=dfa,x='scores',kde=True)
```

```python
sns.histplot(data=dfb,x='prices',kde=True)
```

```python
Q1=dfb['prices'].quantile(0.25)
Q2=dfb['prices'].quantile(0.75)
IQR=Q2-Q1
min_score=max(0,Q1-(1.5*IQR))
max_score=Q3+(1.5*IQR)
outliers=dfb[(dfb['prices']<min_score)|(dfb['prices']>max_score)]
outliers.count()
```
