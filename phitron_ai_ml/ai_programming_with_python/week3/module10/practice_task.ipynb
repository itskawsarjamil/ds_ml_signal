```python
import random
import string
names=["".join(random.choices(string.ascii_uppercase, k=3)) for _ in range(6)]
print(names)
```

```python
marks1=random.sample(range(50,100),6)
print(marks1)
marks2=random.sample(range(50,100),6)
print(marks2)
```

```python
df4=pd.DataFrame({'Name':names,'Marks1':marks1,'Marks2':marks2})
df4

```

```python
df4.drop(columns=['Marks2'],inplace=True)
df4
```
