```python
def introduce(name,age):
    print(f"My name is {name} and I am {age} years old")

introduce("kj",1)
```

```python
def batch_generator(data,batch_size):
    for i in range(0,len(data),batch_size):
        yield data[i:i+batch_size]

x=batch_generator([1,2,3,4,5,6,7,8],3)
print(next(x))
print(next(x))
```
