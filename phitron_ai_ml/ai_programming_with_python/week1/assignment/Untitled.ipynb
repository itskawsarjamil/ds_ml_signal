```python
def compute(a,b):
    try:
        res = (a**2+b**2)/(a-b)
    except ZeroDivisionError:
      return ("you can't divide a number by 0")
    else:
        return res


values=input()
values=values.split()
values=[int(x) for x in values]
a=values[0]
b=values[1]
print(compute(a,b))
```

```python
from functools import reduce
import string
with open('data.txt','r') as file:
    sentence=file.readlines()
    print(sentence)
    
    sentence=list(map(str.strip,sentence))
    # sentence=list(map(lambda st:st.strip(),sentence))
    print((sentence))
    # sentence=list(map(str.lower,sentence))
    # print(sentence)
    # # sentence=reduce(lambda x,y:x+" "+y,sentence)
       
    sentence=" ".join(sentence)
    print(sentence) 
    sentence=sentence.lower()
    print(sentence)
    # sentence=sentence.translate(str.maketrans('','',string.punctuation))
    
    for ch in string.punctuation:
        if ch in sentence:
            sentence=sentence.replace(ch,"")
    print(sentence)
    word_list=list(sentence.split())
    print(word_list)
    maxCount=0
    maxString=""
    frequency={}
    for word in word_list:
        if word in frequency:
            frequency[word]=frequency[word]+1
            if frequency[word]>maxCount:
                maxCount=frequency[word]
                maxString=word
            
        else:
            if maxCount==0:
                maxCount=1
                maxString=word
            frequency[word]=1
    print(maxString)
```

```python
number=input()
i=0;
j=len(number)-1
flag=0
while i<j:
    if number[i]!=number[j]:
        print("Not Mirror Number")
        flag=1
        break
    else:
        i=i+1
        j=j-1
if flag==0:
    print("Mirror Number")

```
