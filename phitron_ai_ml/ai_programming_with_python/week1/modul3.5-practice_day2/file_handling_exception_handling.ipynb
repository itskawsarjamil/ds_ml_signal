```python
def create_file(filename,content):
    with open(filename,'w') as wfile:
        wfile.write(content)

create_file("notes.txt","Hello, this is my first file.")
```

```python
def create_file(filename,content):
    with open(filename,'a') as wfile:
        wfile.write(content)

create_file("data.txt","Apple\nBanana\nCherry\n")
```

```python
def read_lines(filename):
    with open(filename,'r') as rfile:
        lines=rfile.readlines()
        for i,line in enumerate(lines):
            print(f"{i}: {line}")
read_lines("data.txt")
```

```python
def create_file(filename,content):
    with open(filename,'a') as wfile:
        wfile.write(content)

create_file("hello.txt","Hello, World!")
```

```python
def read_from_position(filename,position):
    with open(filename,'r') as rfile:
        rfile.seek(position)
        print(rfile.read())

read_from_position("hello.txt",7)
```

```python
def safe_read(filename):
    file=None
    try:
        file=open(filename,"r")
    except Exception as e:
        print(e)
    else:
        print(file.read())
    finally:
        if file:
            file.close()
        print("Operation complete")

safe_read("missing.txt")
```
