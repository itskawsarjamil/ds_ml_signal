```python
def adder(a,b):
    """
    add two numbers
    a->int
    b-> int

    """
    sum=a+b
    return sum
```

```python
def adder(*args):
    for i in args:
        print(i)
adder(10,20,3,4,5)
```

```python
def adder(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}: {value}")

adder(a=2,b=3,c=4,n_estim=10)
```

```python
lst=[x for x in range(500)]

def data_loader(chunk_size,lst):
    for i in range(0,len(lst),chunk_size):
        yield lst[i:i+chunk_size]

x=data_loader(5,lst)
print(next(x))
```
